```
import numpy as np
from xtheta.montecarlo.uncertainty import monte_carlo_sensitivity
from xtheta.experiments.benchmarks import get_micius_params

print("--- X-Theta V2 Monte Carlo Uncertainty ---")

micius = get_micius_params()
# Add some hypothetical uncertainty to theta (0.1%)
theta_std = micius['theta'] * 0.001

res = monte_carlo_sensitivity(
    micius['mass'], micius['theta'], micius['r_emit'], micius['r_det'],
    theta_std=theta_std,
    n_samples=500
)

print(f"Micius Simulation Results ({res['samples']} samples):")
print(f"Phi_rel: {res['phi_mean']:.2e} +/- {res['phi_std']:.2e}")
print(f"S_max:   {res['s_max_mean']:.6f} +/- {res['s_max_std']:.2e}")
```
