# 04 Monte Carlo Uncertainty Analysis

This notebook propagates coordinate uncertainties through the X-Theta phase prediction model.

> **Scientific Scope:** This analysis supports the null prediction for Earth-orbit scales by showing that even with coordinate uncertainties, the predicted X-Theta correction remains orders of magnitude below current CHSH sensitivity.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
```

```python
import numpy as np
from xtheta.montecarlo.uncertainty import monte_carlo_sensitivity
from xtheta.experiments.benchmarks import get_micius_params

print("--- X-Theta V2 Monte Carlo Uncertainty ---")

micius = get_micius_params()
# Add some hypothetical uncertainty to theta (0.1%)
theta_std = micius['theta'] * 0.001

res = monte_carlo_sensitivity(
    micius['mass'], micius['theta'], micius['r_emit'], micius['r_det'],
    theta_std=theta_std,
    n_samples=500
)

print(f"Micius Simulation Results ({res['samples']} samples):")
print(f"Phi_rel: {res['Phi_mean']:.2e} +/- {res['Phi_std']:.2e}")
print(f"S_max:   {res['s_max_mean']:.6f} +/- {res['s_max_std']:.2e}")
```

## Conclusion

Monte Carlo propagation confirms that coordinate uncertainties do not bring the X-Theta effect into the measurable regime for Earth-orbit experiments.

**Claim Classification:** This result is **simulation/benchmark**. It quantifies the robustness of the null-prediction in the satellite regime.
