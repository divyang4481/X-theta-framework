# Entanglement Lensing: Correlation Surfaces

This notebook visualizes the deformation of the two-qubit correlation tensor under the X-Theta V2 relational phase evolution. 

We distinguish between two types of surfaces:
1. **Direct Correlation-Strength Ellipsoid**: Shows the actual observable correlation magnitude along each axis. Radii: $r_x = r_y = |\cos(2\Phi)|, r_z = 1$.
2. **Dual Response Ellipsoid**: The inverse surface defined by $v^T(T^TT)v=1$. Radii: $r_x = r_y = 1/|\cos(2\Phi)|, r_z = 1$.

```python
import numpy as np
import matplotlib.pyplot as plt
from xtheta.visualization.plots import plot_correlation_strength_ellipsoid, plot_dual_response_ellipsoid

print("--- X-Theta V2 Entanglement Lensing Surfaces ---")
```

## 1. Case $\Phi = 0$ (Isotropic Bell Singlet)

Both surfaces should be unit spheres.

```python
phi = 0.0

fig = plt.figure(figsize=(12, 6))
ax1 = fig.add_subplot(121, projection='3d')
plot_correlation_strength_ellipsoid(phi, ax=ax1)
ax1.set_title(f"Strength (Isotropic) $\Phi={phi}$")

ax2 = fig.add_subplot(122, projection='3d')
plot_dual_response_ellipsoid(phi, ax=ax2)
ax2.set_title(f"Dual Response (Isotropic) $\Phi={phi}$")

plt.tight_layout()
plt.savefig('lensing_phi_0.png')
plt.show()
```

## 2. Case $\Phi = 0.4$ (Anisotropic Lensing)

- The **Strength Ellipsoid** contracts in the X-Y plane (transverse directions).
- The **Dual Response Ellipsoid** expands in the X-Y plane (where correlation is weaker).

```python
phi = 0.4

fig = plt.figure(figsize=(12, 6))
ax1 = fig.add_subplot(121, projection='3d')
plot_correlation_strength_ellipsoid(phi, ax=ax1)
ax1.set_title(f"Strength (Contracted) $\Phi={phi}$")

ax2 = fig.add_subplot(122, projection='3d')
plot_dual_response_ellipsoid(phi, ax=ax2)
ax2.set_title(f"Dual Response (Expanded) $\Phi={phi}$")

plt.tight_layout()
plt.savefig('lensing_phi_04.png')
plt.show()
```

## Conclusion

The direct correlation-strength ellipsoid shows actual observable correlation magnitudes. The dual response ellipsoid is the inverse surface $v^T(T^TT)v=1$ and expands in directions where correlation strength weakens.
