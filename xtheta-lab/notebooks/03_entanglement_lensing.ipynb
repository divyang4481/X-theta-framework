```
import numpy as np
import matplotlib.pyplot as plt
from xtheta.visualization.plots import plot_correlation_ellipsoid

print("--- X-Theta V2 Entanglement Lensing ---")

# High phase case (e.g. near Neutron Star)
phi_high = 0.4
fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')
plot_correlation_ellipsoid(phi_high, ax=ax)
plt.savefig('entanglement_lensing_high.png')

# Zero phase case (Standard QM)
phi_zero = 0.0
fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')
plot_correlation_ellipsoid(phi_zero, ax=ax)
plt.savefig('entanglement_lensing_zero.png')

print("Plots saved: entanglement_lensing_high.png, entanglement_lensing_zero.png")
```
