```python
import numpy as np
import pandas as pd
from xtheta.experiments.benchmarks import run_benchmark_scenarios
from xtheta.quantum.engine import compute_chsh_max, evolve_state, compute_correlation_tensor

print("--- X-Theta V2 Benchmark Scenarios ---")

results = run_benchmark_scenarios()

for r in results:
    phi = r['phi_rel']
    psi = evolve_state(phi)
    T = compute_correlation_tensor(psi)
    s_max = compute_chsh_max(T)
    
    r['S_max'] = s_max
    r['delta_S'] = abs(s_max - 2*np.sqrt(2))

df = pd.DataFrame(results)
print(df[['name', 'phi_rel', 'S_max', 'delta_S']])
```
